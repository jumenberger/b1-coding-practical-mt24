```python
# Unit1: Importing Necessary Modules
import sys
import os

# Importing Project Root to Python Path
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.append(project_root)

# Imporrting from uuv_mission
from uuv_mission.dynamic import Submarine, Mission, ClosedLoop
from uuv_mission.controller import PDController
import matplotlib.pyplot as plt

print("Import successful!")
print(f"Project Root: {project_root}")
```

```python
# Unit2: Creating Controller and System Objects
# Provided values
controller = PDController(Kp=0.15, Kd=0.6)
sub = Submarine()
closed_loop = ClosedLoop(sub, controller)

print("Creation successful!")
```

```python
# Unit3: Loading Mission Data
mission = Mission.from_csv("../data/mission.csv")
print(f"Load successful！")
print(f"Duration: {len(mission.reference)} steps")
```

```python
# Unit4: Run Simulation
trajectory = closed_loop.simulate_with_random_disturbances(mission, variance=0.5)
print("Simulation successful！")

# Unit5: Plot Results
trajectory.plot_completed_mission(mission)
```
