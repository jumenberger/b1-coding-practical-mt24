```python
import numpy as np
import uuv_mission.control
from uuv_mission.dynamic import Submarine, ClosedLoop, Mission
```

```python
# Controller Test

def dummy_controller():
    return 0.0

submarine = Submarine()
closed_loop = ClosedLoop(submarine, dummy_controller)

mission = Mission(
    reference = np.array([0,1,2,3,4]),
    cave_height = np.array([-5,-5,-5,-5,-5]),
    cave_depth = np.array([10,10,10,10,10])
)

disturbances = np.zeros(5)

trajectory = closed_loop.simulate(mission, disturbances)

print(trajectory.position)



```
